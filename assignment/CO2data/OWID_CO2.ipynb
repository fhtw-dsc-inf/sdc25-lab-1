# Import CO2 data from
Our World in Data - CO2 Emissions: Provides a comprehensive dataset on global CO2 emissions, including emissions by country and by fuel type.
https://ourworldindata.org/co2-emissions

Datasource: https://github.com/owid/co2-data

```python
import pandas as pd

# URL of the CO2 dataset CSV file
data_url = 'https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv'

# Download and read the dataset into a pandas DataFrame
df = pd.read_csv(data_url)

# Display the first few rows of the DataFrame
print(df.head())

# Save this DataFrame to a local CSV file
df.to_csv('owid_co2_data.csv', index=False)
```
