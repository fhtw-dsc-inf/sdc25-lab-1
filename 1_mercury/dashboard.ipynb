```python
import yfinance as yf
import mplfinance as mpf # matplotlib but optimized for financial data
import mercury as mr
```

# Historic stock price 📈

```python

# add select widget
ticker = mr.Select(value="TSLA", choices=['TSLA', 'AMZN', 'COIN', 'MSFT', 'PLTR', 'SNOW', 'NFLX'], label="select stock")
```

```python
print(f"Selected ticker: {ticker.value}")
```

```python
history = yf.Ticker(ticker.value).history(period="3mo")
```

```python
print(f"3 months low: {history.Low.min()}")
print(f"3 months high: {history.High.max()}")
```

```python
history.tail()
```

```python
mpf.plot(history, type='candle', mav=(7),figratio=(18,10))
```
